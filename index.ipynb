```python
def search_laws(query, field="rule_id", data=jsonObject):
    results = []
    query_lower = str(query).lower()
    
    for rule in data:
        field_value = str(rule.get(field, "")).lower()
        if query_lower in field_value:
            results.append(rule)
            
    return results
```

```python
matches = search_laws("UK-TORT-002", field="rule_id")
for m in matches:
    print(f"Found: {m['rule_id']} - {m['claim_type']}")

print("\n")

print("Search by Claim Type")
tort_matches = search_laws("UK-TORT-002", field="period_value")
for m in tort_matches:
    print(f"Time: {m['period_value'], "years" } | Claim: {m['claim_type']} | Citation: {m['citation']}")
```

```python


user_input = input("Enter the claim or offense (e.g., contract): ")
start_date = input("Enter start date of the cause of action (YYYY-MM-DD): ")
start_date2 = datetime.strptime(start_date, "%Y-%m-%d")

results = search_laws(user_input, field="claim_type")

if len(results) == False:
    print("No matching statute found for that offense.")
elif len(results) == True:
    rule = results[0]
    print(f"\n[Match Found] {rule['rule_id']} - {rule['claim_type']} ({rule['jurisdiction']})")
    print(f"Statutory Citation: {rule['citation']}")
    
    val = rule["period_value"]
    unit = rule["period_unit"]
    
    if val ==0:
        print("This claim has no standard statute of limitations (not time-barred).")
    else:
        if unit == "years":
            days_to_add = val * 365  
        elif unit == "months":
            days_to_add = val * 30
        else:
            days_to_add = val
            
        deadline = start_date + timedelta(days=days_to_add)
        
        print(f"Limitation Period: {val} {unit}")
        print(f"Action Expires On (Deadline): {deadline.strftime('%Y-%m-%d')}")
else:
    print(f"\nFound {len(results)} matching rules. Please be more specific or select one:")
    for idx, r in enumerate(results):
        print(f"[{idx}] {r['rule_id']} - {r['claim_type']} in {r['jurisdiction']} ({r['period_value']} {r['period_unit']})")
```
